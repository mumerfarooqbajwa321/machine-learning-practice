# Nominal Data Encoding
**One hot encoding is used for nominal data encoding**

```python
import pandas as pd
```

```python
df = pd.read_csv("/kaggle/input/datasets/volodymyrgavrysh/bank-marketing-campaigns-dataset/bank-additional-full.csv", sep=';')
```

```python
df
```

```python
#nominal encoding with pandas

df_ohe=pd.get_dummies(df, columns= ['contact','marital'])
```

```python
df_ohe.sample(5)
```

```python
#k-1 one hot encoding (remove collinearity ie dropping one columns)
df_ohe_k_1 = pd.get_dummies(df, columns=['marital', 'contact'], drop_first=True)
```

```python
df_ohe_k_1.sample(5)
```

# One Hot Encoding With SciKit Learn

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe= OneHotEncoder(drop='first')
```

```python
ohe.fit(df[["contact","marital"]])
```

```python
encoded_df = ohe.transform(df[['contact','marital']])
```

```python
 encoded_df = encoded_df.toarray()
 encoded_df
```

```python
counts=df["month"].value_counts()
counts
```

```python
#i want months with less than 1000 existences ie value counts, i want gather all of them in a new column named "uncommon"
```

```python
threshold = 1000
uncommon_months=counts [ counts<threshold].index
uncommon_months
```

```python
pd.get_dummies(df['month'].replace(uncommon_months, "uncommon"), drop_first= True)
```
