# **Gradient Descent Implementaion**

```python
class GDRegressor:

    def __init__ (self, learning_rate=0.001, epochs=100):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.coef_ = None
        self.intercept_ = None

    def fit(self,X_train, y_train):
        #changing pandas dataframe into arrays
        X_train = X_train.values if hasattr(X_train, 'values') else np.asarray(X_train)
        y_train = y_train.values if hasattr(y_train, 'values') else np.asarray(y_train)
        
        self.intercept_ = 0
        self.coef_ = np.array([1 for i in range (X_train.shape[1])])

        # intercept (beta-not) = intercept - learning_rate (eeta) * slope of loss function ( like mse)
        
        # slope can be calculated by derivative
        
        # y_hat is predicted y (with previous weights)
        for k in range(self.epochs):
            y_hat = np.dot( X_train, self.coef_.reshape(self.coef_.shape[0],1) ) + self.intercept_   # self.coef_ 1D, X_train 2D, so y-hat is 1D ie predicted y for whole X_train instead of one row only
            
            y_hat = y_hat.reshape(-1)
            
            error_difference = y_train - y_hat
            
            der_intercept = -2* np.mean(error_difference)
            
            # updating intercept
            self.intercept_ = self.intercept_ - self.learning_rate * der_intercept
            
            # der_Beta_one ie. slope with respect to X1 of any row 
            #  der_beta_one = 2 / n  * sum ( (y - y_hat) * X1)
            #  generally for each beta/weight:
            # der_coef_[i] = 2 * np.mean((y_train- y_hat)*X_train[i])

            der_coef_ = np.dot(error_difference , X_train) # yet to calucalte its 2 * mean 
            der_coef_ = -2/X_train.shape[0] * der_coef_
            
            #updating coef_ weights
            self.coef_ = self.coef_ - self.learning_rate * der_coef_

    def predict(self,X_test):
        X_test = X_test.values if hasattr(X_test, 'values') else np.asarray(X_test)
        return np.dot(X_test, self.coef_) + self.intercept
    
```
