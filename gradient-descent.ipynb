```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test= train_test_split(df.iloc[:, :-1], df.iloc[:, -1], test_size= 0.2, random_state =42)

X_train, X_test, y_train, y_test
```

```python
print(type(X_train))
print(type(X_test))
print(type(y_train))
print(type(y_test))
```

```python
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder

from sklearn.preprocessing import StandardScaler


from sklearn.compose import ColumnTransformer

from sklearn.pipeline import Pipeline


rtf= ColumnTransformer([('ohe', OneHotEncoder(drop='first', handle_unknown='error', sparse_output= False ), ['sex', 'region']),
                       ('oe', OrdinalEncoder(categories=[['no','yes']]), ['smoker'] ),],
                      remainder='passthrough')

pipeline = Pipeline ([
    ('encoding', rtf)
    #('scaler', StandardScaler())
])
```

```python
pipeline.fit(X_train)
```

```python
X_train_trf = pipeline.transform(X_train)
X_test_trf = pipeline.transform (X_test)
```
