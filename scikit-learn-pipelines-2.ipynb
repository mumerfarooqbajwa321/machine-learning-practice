```python
import pandas as pd
df = pd.read_csv("/kaggle/input/datasets/organizations/uciml/adult-census-income/adult.csv")
```

```python
df = df.drop(columns=["education.num","relationship"])
df.sample(5)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test, Y_train, Y_test = train_test_split(df.drop(columns=['income']), df['income'], test_size=0.2, random_state=42)
X_train
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder , OrdinalEncoder
from sklearn.pipeline import Pipeline , make_pipeline
```

```python
trf1 = ColumnTransformer([(
        "education_oe",OrdinalEncoder (categories=[["Preschool","1st-4th",'5th-6th','7th-8th','9th','10th','HS-grad','11th','12th','Some-college','Prof-school','Assoc-acdm','Assoc-voc','Bachelors','Masters','Doctorate']], handle_unknown='error'),[3]
    )],
remainder= 'passthrough')
#array(['HS-grad', 'Bachelors', 'Some-college', '11th', '5th-6th',
#       'Assoc-voc', 'Masters', '9th', 'Doctorate', 'Prof-school',
#       '7th-8th', '10th', '12th', '1st-4th', 'Assoc-acdm', 'Preschool'],
#      dtype=object)
```

```python
trf2 = ColumnTransformer ([(
'ohe' , OneHotEncoder(drop='first',sparse_output=False, handle_unknown = 'ignore'), [2,4,5,6,7,11]
)], 
remainder= 'passthrough')
```

```python
from sklearn.tree import DecisionTreeClassifier

trf3 = DecisionTreeClassifier()
```

```python
pipeline = Pipeline([
    ('trf1', trf1),
    ('trf2', trf2),
    ('trf3', trf3)
])
```

```python
X_train['education'].unique()
```

```python
from sklearn.preprocessing import LabelEncoder
le= LabelEncoder()
```

```python
pipeline.fit(X_train, Y_train)
```

```python
predicted_Y= pipeline.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(Y_test, predicted_Y)
```
